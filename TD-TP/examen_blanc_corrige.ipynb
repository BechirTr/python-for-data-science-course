## Exercice 4. Tracer une recherche dichotomique

Chapitre 3 · recherche et complexité

```python
def chercher(valeurs, cible):
    gauche, droite = 0, len(valeurs) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if valeurs[milieu] == cible:
            return milieu
        if valeurs[milieu] < cible:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return -1

```

1. Tracer l’appel chercher(valeurs, 13) et donner son résultat.
2. Combien d’itérations ont lieu ici ?
3. Quelle condition sur valeurs permet cette recherche ? Donner le coût en temps au pire et l’espace auxiliaire, pour n valeurs.

### Correction

La recherche renvoie l’indice 4 après trois tours. La liste doit être triée selon l’ordre de comparaison. Chaque tour retire environ la moitié de l’intervalle : O(log n) au pire, O(1) d’espace auxiliaire. Le nombre de comparaisons élémentaires dépend des branches, il ne faut pas le confondre avec les trois tours.

gauche | droite | milieu | valeur lue | Décision
--- | --- | --- | --- | ---
0 | 6 | 3 | 10 | gauche = 4
4 | 6 | 5 | 16 | droite = 4
4 | 4 | 4 | 13 | renvoyer 4

````python
def chercher(valeurs, cible):
    gauche, droite = 0, len(valeurs) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if valeurs[milieu] == cible:
            return milieu
        if valeurs[milieu] < cible:
            gauche = milieu + 1
        else:
            droite = milieu - 1
    return -1
print(chercher([1,4,7,10,13,16,19],13))

assert chercher([],3) == -1
assert chercher([1,4,7,10,13,16,19],13) == 4
assert chercher([1,4,7],6) == -1
````

## Exercice 6. Corriger une recherche du dernier indice

Chapitres 1 à 3 · contrat et retour prématuré

```python

    indice = -1
    for i in range(len(valeurs)):
        if valeurs[i] == cible:
            return i
    return indice
```

1. Le contrat demande le dernier indice égal à cible, ou -1 si elle est absente. Donner un contre-exemple au programme.
2. Corriger le programme sans inverser ni trier la liste.
3. Tester une liste vide, une cible absente et plusieurs occurrences. Donner la complexité.

### Correction

Pour [4, 2, 4] et la cible 4, le programme renvoie 0 au lieu de 2. Il faut mémoriser chaque indice trouvé et attendre la fin du parcours. Le coût est O(n) en temps et O(1) en espace auxiliaire.

````python
def dernier_indice(valeurs, cible):
    indice = -1
    for i in range(len(valeurs)):
        if valeurs[i] == cible:
            indice = i
    return indice
print(dernier_indice([4,2,4],4))

assert dernier_indice([],4) == -1
assert dernier_indice([1,2],4) == -1
assert dernier_indice([4,2,4],4) == 2
````


## Exercice 8. Corriger une propriété et sa validation

Chapitre 4 · constructeur, getter et setter

```python
class Batterie:
    def __init__(self, capacite):
        self._capacite = capacite

    @property
    def capacite(self):
        return self._capacite

    @capacite.setter
    def capacite(self, valeur):
        if valeur < 0:
            raise ValueError("capacite negative")
        self.capacite = valeur
```

1. Le contrat interdit une capacité négative, dès la création. Pourquoi Batterie(-1) ne respecte-t-il pas ce contrat ?
2. Pourquoi l’affectation b.capacite = 5 provoque-t-elle une récursion ?
3. Corriger les deux lignes concernées. Vérifier qu’une affectation invalide laisse la valeur précédente inchangée.

### Correction

Le constructeur écrit directement _capacite et contourne le setter. Le setter réaffecte capacite, ce qui le rappelle indéfiniment. Le constructeur doit passer par self.capacite, puis le setter validé doit écrire self._capacite. Une valeur négative déclenche ValueError avant toute modification.

````python
class Batterie:
    def __init__(self, capacite):
        self.capacite = capacite
    @property
    def capacite(self):
        return self._capacite
    @capacite.setter
    def capacite(self, valeur):
        if valeur < 0:
            raise ValueError("capacite negative")
        self._capacite = valeur

batterie = Batterie(10)
batterie.capacite = 5
print(batterie.capacite)

assert Batterie(0).capacite == 0
try:
    Batterie(-1)
    raise AssertionError("validation absente")
except ValueError:
    pass
try:
    batterie.capacite = -2
    raise AssertionError("validation absente")
except ValueError:
    pass
assert batterie.capacite == 5
````

## Problème 1. Regrouper les répétitions consécutives

On veut résumer les groupes de caractères consécutifs d’un texte. La sortie est une liste de couples (caractère, nombre). Deux groupes séparés restent séparés, même si leur caractère est le même. Les caractères sont sensibles à la casse. Une chaîne vide donne une liste vide.

**Exemples** : "aaabbca" donne [("a", 3), ("b", 2), ("c", 1), ("a", 1)]. "Aaa" donne [("A", 1), ("a", 2)].

1. Décrire ce que l’on mémorise pour le groupe courant.
2. Écrire un pseudocode pour compresser(texte), sans bibliothèque de comptage.
3. Tracer "aabb" en montrant le groupe courant et la sortie.
4. Donner un test vide, un test sans répétition et les coûts en temps et en mémoire.

### Pseudocode

```text
FONCTION compresser(texte)
    SI texte est vide : RENVOYER []
    sortie <- []
    caractere <- texte[0]
    nombre <- 1
    POUR chaque position i de 1 à longueur(texte)-1
        SI texte[i] = caractere
            nombre <- nombre + 1
        SINON
            ajouter (caractere, nombre) à sortie
            caractere <- texte[i]
            nombre <- 1
    ajouter (caractere, nombre) à sortie
    RENVOYER sortie
```

### Complexité et points à vérifier

Un seul parcours : O(n) en temps. La sortie contient g groupes : O(g), donc O(n) au pire. En dehors de la sortie, quelques variables suffisent : O(1). Le dernier groupe doit être ajouté après la boucle.

### Trace

Étape | Groupe courant | Sortie
--- | --- | ---
Initialisation | a, 1 | []
Deuxième a | a, 2 | []
Premier b | b, 1 | [('a', 2)]
Deuxième b | b, 2 | [('a', 2)]
Fin | b, 2 | [('a', 2), ('b', 2)]

````python
def compresser(texte):
    if texte == "":
        return []
    sortie = []
    caractere, nombre = texte[0], 1
    for i in range(1, len(texte)):
        if texte[i] == caractere:
            nombre += 1
        else:
            sortie.append((caractere, nombre))
            caractere, nombre = texte[i], 1
    sortie.append((caractere, nombre))
    return sortie

assert compresser("") == []
assert compresser("aaabbca") == [("a",3),("b",2),("c",1),("a",1)]
assert compresser("aabb") == [("a",2),("b",2)]
assert compresser("abc") == [("a",1),("b",1),("c",1)]
assert compresser("Aaa") == [("A",1),("a",2)]
````

## Problème 2. La plus longue série de jours positifs

Une liste contient le bilan quotidien d’une activité. Un jour est positif si sa valeur est strictement supérieure à zéro. On cherche la longueur de la plus longue série de jours positifs consécutifs. Un zéro ou une valeur négative coupe la série. Une liste vide donne zéro.

**Exemples** : [2, 1, 0, 3, 4, 5, -1, 2] donne 3, grâce à [3, 4, 5]. [-1, 0] donne 0.

1. Nommer les deux compteurs utiles et expliquer leur rôle.
2. Écrire le pseudocode de plus_longue_serie(bilans).
3. Tracer l’exemple en montrant la série courante et le meilleur résultat.
4. Prévoir le cas où tous les jours sont positifs et donner les complexités.

### Pseudocode

```text
FONCTION plus_longue_serie(bilans)
    courante <- 0
    meilleure <- 0
    POUR chaque valeur dans bilans
        SI valeur > 0
            courante <- courante + 1
            SI courante > meilleure
                meilleure <- courante
        SINON
            courante <- 0
    RENVOYER meilleure
```

### Complexité et points à vérifier

Chaque valeur est lue une fois : O(n) en temps. Deux compteurs indépendants de n : O(1) en espace auxiliaire. La meilleure série doit rester mémorisée lorsque la série courante revient à zéro.

### Trace

Valeur | courante | meilleure
--- | --- | ---
2 | 1 | 1
1 | 2 | 2
0 | 0 | 2
3 | 1 | 2
4 | 2 | 2
5 | 3 | 3
-1 | 0 | 3
2 | 1 | 3

````python
def plus_longue_serie(bilans):
    courante = 0
    meilleure = 0
    for valeur in bilans:
        if valeur > 0:
            courante += 1
            if courante > meilleure:
                meilleure = courante
        else:
            courante = 0
    return meilleure

assert plus_longue_serie([]) == 0
assert plus_longue_serie([-1,0]) == 0
assert plus_longue_serie([1,2,3]) == 3
assert plus_longue_serie([2,1,0,3,4,5,-1,2]) == 3
````

## Problème 3. Fusionner deux relevés déjà triés

Deux appareils produisent chacun une liste d’entiers triée par ordre croissant. On veut une nouvelle liste triée contenant toutes les valeurs, y compris les doublons. Les listes d’entrée restent inchangées. Elles peuvent être vides. On n’utilise ni sorted ni sort.

**Exemples** : [1, 4, 4] et [2, 4, 6] donnent [1, 2, 4, 4, 4, 6]. [] et [2] donnent [2].

1. Expliquer comment choisir la prochaine valeur avec deux indices.
2. Écrire le pseudocode de fusionner(gauche, droite). En cas d’égalité, prendre d’abord la valeur de gauche.
3. Tracer l’exemple, puis expliquer comment terminer lorsqu’une liste est épuisée.
4. Donner les coûts pour n et m valeurs. Distinguer la sortie et la mémoire auxiliaire.

### Pseudocode

```text
FONCTION fusionner(gauche, droite)
    i <- 0 ; j <- 0 ; sortie <- []
    TANT QUE i < longueur(gauche) ET j < longueur(droite)
        SI gauche[i] <= droite[j]
            ajouter gauche[i] à sortie
            i <- i + 1
        SINON
            ajouter droite[j] à sortie
            j <- j + 1
    TANT QUE i < longueur(gauche)
        ajouter gauche[i] à sortie
        i <- i + 1
    TANT QUE j < longueur(droite)
        ajouter droite[j] à sortie
        j <- j + 1
    RENVOYER sortie
```

### Complexité et points à vérifier

Chaque élément rejoint la sortie une fois : O(n + m) en temps. La sortie occupe O(n + m). Les deux indices n’exigent que O(1) d’espace auxiliaire. Les doublons restent présents.

### Trace

Valeur ajoutée | i après ajout | j après ajout | Origine
--- | --- | --- | ---
1 | 1 | 0 | gauche
2 | 1 | 1 | droite
4 | 2 | 1 | gauche, égalité
4 | 3 | 1 | gauche, égalité
4 | 3 | 2 | reste droite
6 | 3 | 3 | reste droite

````python
def fusionner(gauche, droite):
    i, j = 0, 0
    sortie = []
    while i < len(gauche) and j < len(droite):
        if gauche[i] <= droite[j]:
            sortie.append(gauche[i])
            i += 1
        else:
            sortie.append(droite[j])
            j += 1
    while i < len(gauche):
        sortie.append(gauche[i])
        i += 1
    while j < len(droite):
        sortie.append(droite[j])
        j += 1
    return sortie

assert fusionner([],[]) == []
assert fusionner([],[2]) == [2]
assert fusionner([1,4,4],[2,4,6]) == [1,2,4,4,4,6]
assert fusionner([-2,3],[-1,3]) == [-2,-1,3,3]
````

## Problème 4. Quels postes reçoivent le message ?

Un réseau orienté indique à quels postes un poste peut transmettre un message. La clé d’un dictionnaire est un poste et sa valeur est la liste de ses destinataires directs. On veut renvoyer les postes accessibles depuis le départ, chacun une seule fois, départ inclus. Les cycles ne doivent pas provoquer une boucle infinie. Chaque destinataire apparaît aussi comme clé. Si le départ est absent, renvoyer [].

**Exemples** : A transmet à B et C. B transmet à D. C transmet à D. D transmet à B. E transmet à F. F ne transmet à personne. Depuis A, on doit obtenir [A, B, C, D] en traitant les voisins dans l’ordre donné.

1. Dessiner les six postes et les flèches.
2. Écrire un pseudocode avec une liste de postes à traiter, un indice courant et un ensemble de postes déjà découverts.
3. Marquer un poste dès son ajout à la liste. Tracer le parcours depuis A.
4. Pourquoi E et F ne figurent-ils pas dans le résultat ? Donner les complexités.

### Pseudocode

```text
FONCTION accessibles(graphe, depart)
    SI depart n'est pas une cle du graphe : RENVOYER []
    a_traiter <- [depart]
    vus <- ensemble contenant depart
    i <- 0
    TANT QUE i < longueur(a_traiter)
        sommet <- a_traiter[i]
        POUR chaque voisin dans graphe[sommet]
            SI voisin n'est pas dans vus
                ajouter voisin à vus
                ajouter voisin à a_traiter
        i <- i + 1
    RENVOYER a_traiter
```

### Complexité et points à vérifier

C’est un parcours en largeur. Avec V postes et E liens dans la partie atteignable, le coût moyen est O(V + E) en temps et O(V) en mémoire, sous le coût moyen O(1) des ensembles. L’indice évite les décalages de pop(0). E et F ne sont reliés à aucun chemin partant de A.

### Trace

Poste traité | Nouveaux postes | Liste après traitement
--- | --- | ---
A | B, C | [A, B, C]
B | D | [A, B, C, D]
C | aucun | [A, B, C, D]
D | aucun | [A, B, C, D]

````python
def accessibles(graphe, depart):
    if depart not in graphe:
        return []
    a_traiter = [depart]
    vus = {depart}
    i = 0
    while i < len(a_traiter):
        sommet = a_traiter[i]
        for voisin in graphe[sommet]:
            if voisin not in vus:
                vus.add(voisin)
                a_traiter.append(voisin)
        i += 1
    return a_traiter

graphe = {"A":["B","C"],"B":["D"],"C":["D"],
          "D":["B"],"E":["F"],"F":[]}
assert accessibles(graphe,"A") == ["A","B","C","D"]
assert accessibles(graphe,"E") == ["E","F"]
assert accessibles(graphe,"absent") == []
assert accessibles({"A":["A"]},"A") == ["A"]
````

## Problème 5. Concevoir un petit stock de matériel

Une association gère des quantités de matériel par nom d’article. Concevoir une classe Stock. Chaque instance commence vide et possède ses propres données. disponible(article) renvoie la quantité, ou 0 si l’article est inconnu. ajouter(article, quantite) augmente le stock, en créant l’article si nécessaire. retirer(article, quantite) renvoie True et diminue le stock si la quantité suffit. Sinon, elle renvoie False et ne change rien. Les quantités reçues sont des entiers. Une quantité inférieure ou égale à zéro déclenche ValueError avant toute modification.

**Exemples** : Après ajouter("cable", 3), disponible("cable") vaut 3. retirer("cable", 2) renvoie True et laisse 1. retirer("cable", 2) renvoie False et laisse encore 1. Un second stock reste vide.

1. Choisir un attribut et sa structure de données. Justifier ce choix.
2. Écrire le pseudocode du constructeur et des trois méthodes.
3. Tracer les trois opérations de l’exemple en montrant l’état après chaque appel.
4. Proposer des tests pour une clé absente, un retrait refusé, une quantité invalide et deux instances indépendantes.

### Pseudocode

```text
CLASSE Stock
    CONSTRUCTEUR
        quantites <- nouveau dictionnaire vide
    METHODE disponible(article)
        RENVOYER quantites.get(article, 0)
    METHODE ajouter(article, q)
        SI q <= 0 : DECLENCHER ValueError
        quantites[article] <- disponible(article) + q
    METHODE retirer(article, q)
        SI q <= 0 : DECLENCHER ValueError
        SI disponible(article) < q : RENVOYER Faux
        quantites[article] <- disponible(article) - q
        RENVOYER Vrai
```

### Complexité et points à vérifier

Le dictionnaire associe un article à sa quantité. Chaque méthode a un coût moyen O(1) pour des clés de taille bornée, sous les coûts habituels du hachage. La mémoire est O(k) pour k articles mémorisés. Le dictionnaire doit être créé dans __init__, séparément pour chaque instance.

### Trace

Opération | Retour | Stock de cable
--- | --- | ---
Création | objet | 0
ajouter(cable, 3) | aucun retour utile | 3
retirer(cable, 2) | True | 1
retirer(cable, 2) | False | 1

````python
class Stock:
    def __init__(self):
        self._quantites = {}
    def disponible(self, article):
        return self._quantites.get(article, 0)
    def ajouter(self, article, quantite):
        if quantite <= 0:
            raise ValueError("quantite non positive")
        self._quantites[article] = self.disponible(article) + quantite
    def retirer(self, article, quantite):
        if quantite <= 0:
            raise ValueError("quantite non positive")
        if self.disponible(article) < quantite:
            return False
        self._quantites[article] = self.disponible(article) - quantite
        return True

s1, s2 = Stock(), Stock()
assert s1.disponible("cable") == 0
s1.ajouter("cable",3)
assert s1.retirer("cable",2) is True
assert s1.disponible("cable") == 1
assert s1.retirer("cable",2) is False
assert s1.disponible("cable") == 1 and s2.disponible("cable") == 0
assert s1.retirer("inconnu",1) is False
for q in [0,-1]:
    try:
        s1.ajouter("cable",q)
        raise AssertionError("validation absente")
    except ValueError:
        pass
    try:
        s1.retirer("cable",q)
        raise AssertionError("validation absente")
    except ValueError:
        pass
assert s1.disponible("cable") == 1
````
